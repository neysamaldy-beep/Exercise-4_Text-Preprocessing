```python

import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

URL = "https://lintangdikalasenja.substack.com/p/resiko-terbesar-adalah-tak-mengambil"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/154.0 Safari/537.36"
}

response = requests.get(URL, headers=headers, timeout=30)
response.raise_for_status()

print("Status:", response.status_code)
print("URL:", response.url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

article = soup.select_one(".available-content")

if article is None:
    article = soup.find("article")

if article is None:
    raise ValueError("Isi artikel tidak ditemukan. Coba jalankan ulang atau periksa URL.")

raw_html = str(article)
raw_text = article.get_text(" ", strip=True)

print("Jumlah karakter teks awal:", len(raw_text))
print("\nContoh teks awal:\n")
print(raw_text[:2000])
```

```python
df = pd.DataFrame({
    "Id": [1],
    "text": [raw_html]
})

df.head()
```

## 1. Remove HTML

```python
def remove_html(text):
    return BeautifulSoup(str(text), "html.parser").get_text(" ")

df["rm_html"] = df["text"].apply(remove_html)

print(df["rm_html"].iloc[0][:1500])
```

## 2. Remove Hashtag

```python
def remove_hashtag(text):
    return re.sub(r"#\w+", " ", str(text))

df["rm_hashtag"] = df["rm_html"].apply(remove_hashtag)

print(df["rm_hashtag"].iloc[0][:1500])
```

## 3. Lowercase

```python
df["lower_case"] = df["rm_hashtag"].str.lower()

print(df["lower_case"].iloc[0][:1500])
```

## 4. Remove URL

```python
def remove_url(text):
    return re.sub(r"https?://\S+|www\.\S+", " ", str(text))

df["rm_url"] = df["lower_case"].apply(remove_url)

print(df["rm_url"].iloc[0][:1500])
```

## 5. Remove Punctuation

```python
def remove_punctuation(text):
    # Mempertahankan huruf, angka, dan spasi
    return re.sub(r"[^a-z0-9\s]", " ", str(text))

df["rm_punc"] = df["rm_url"].apply(remove_punctuation)

# Rapikan spasi berlebih
df["rm_punc"] = df["rm_punc"].apply(lambda x: re.sub(r"\s+", " ", x).strip())

print(df["rm_punc"].iloc[0][:1500])
```

## 6. Stemming

```python
factory = StemmerFactory()
stemmer = factory.create_stemmer()

df["stem"] = df["rm_punc"].apply(
    lambda x: stemmer.stem(x) if isinstance(x, str) else ""
)

print(df["stem"].iloc[0][:1500])
```

## 7. Remove Stopwords

```python
stop_factory = StopWordRemoverFactory()
stopwords = set(stop_factory.get_stop_words())

def remove_stopwords(text):
    words = str(text).split()
    return " ".join(word for word in words if word not in stopwords)

df["rm_stopwords"] = df["stem"].apply(remove_stopwords)

df["prepro"] = df["rm_stopwords"].apply(
    lambda x: re.sub(r"\s+", " ", str(x)).strip()
)

print(df["prepro"].iloc[0][:2000])
```

## Perbandingan Sebelum dan Sesudah Preprocessing

```python
comparison = pd.DataFrame({
    "Sebelum Preprocessing": [df["rm_html"].iloc[0]],
    "Sesudah Preprocessing": [df["prepro"].iloc[0]]
})

pd.set_option("display.max_colwidth", 1000)
comparison
```

## Menyimpan Hasil

```python
output = df[["Id", "text", "prepro"]]
output.to_csv("article_preprocessed.csv", index=False)

print("File berhasil dibuat: article_preprocessed.csv")
print("Jumlah dokumen:", len(output))
print("Jumlah karakter hasil preprocessing:", len(output["prepro"].iloc[0]))
```
