```python

!pip install -q gensim umap-learn

import pandas as pd
import numpy as np
import re
import gensim.models
from umap import UMAP
import plotly.express as px

documents = [p.strip() for p in artikel.split('\n') if p.strip()]

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    return text.split()

sentences = [preprocess(doc) for doc in documents]

model = gensim.models.Word2Vec(
    sentences,
    min_count=1,
    vector_size=100,
    window=5,
    sg=0,
    workers=1,
    seed=42
)

model.save('sentimen_tweet.w2v')

model = gensim.models.Word2Vec.load('sentimen_tweet.w2v')

w2v = model.wv

print("10 kata pertama:")
print(w2v.index_to_key[:10])

print("\nUkuran matriks dan vektor:")
print(w2v.vectors.shape, w2v.vector_size)

if 'pakai' in w2v:
    print("\nKata yang mirip dengan 'pakai':")
    print(w2v.similar_by_key('pakai'))
else:
    print("\nKata 'pakai' tidak ditemukan dalam artikel.")

def get_document_vector(tokens, model):
    valid_vectors = [
        model.wv[word] for word in tokens
        if word in model.wv
    ]

    if not valid_vectors:
        return np.zeros(model.vector_size)

    return np.mean(valid_vectors, axis=0)

X_features = np.array([
    get_document_vector(doc, model)
    for doc in sentences
])

df_word2vec = pd.DataFrame(X_features)

print("\nHasil Feature Extraction Word2Vec:")
display(df_word2vec)

print("Ukuran fitur:", X_features.shape)

X = UMAP(
    n_components=2,
    random_state=42
).fit_transform(w2v.vectors)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])
df2['word'] = w2v.index_to_key

fig = px.scatter(
    df2,
    x='umap1',
    y='umap2',
    text='word',
    title='Word2Vec Visualization'
)

fig.update_traces(textposition='top center')
fig.update_layout(height=800)
fig.show()
```

```python

import gensim.models
import numpy as np
import pandas as pd
import plotly.express as px
from umap import UMAP

model_fasttext = gensim.models.FastText(
    sentences,
    min_count=1,
    vector_size=100,
    window=5,
    min_n=3,
    max_n=6,
    sg=1,
    epochs=10,
    workers=1,
    seed=42
)

model_fasttext.save('sentimen_tweet.fasttext')

model_fasttext = gensim.models.FastText.load(
    'sentimen_tweet.fasttext'
)

fastext = model_fasttext.wv

print("10 kata pertama:")
print(fastext.index_to_key[:10])

print("\nUkuran matriks vektor:")
print(fastext.vectors.shape)

def get_document_vector(tokens, model):
    valid_vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]

    if not valid_vectors:
        return np.zeros(model.vector_size)

    return np.mean(valid_vectors, axis=0)

X_features = np.array([
    get_document_vector(doc, model_fasttext)
    for doc in sentences
])

df_fasttext = pd.DataFrame(X_features)

print("\nHasil Feature Extraction FastText:")
display(df_fasttext)

print("Ukuran fitur:", X_features.shape)

X = UMAP(
    n_components=2,
    random_state=42
).fit_transform(fastext.vectors)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])
df2['word'] = fastext.index_to_key

fig = px.scatter(
    df2,
    x='umap1',
    y='umap2',
    text='word',
    title='FastText Visualization'
)

fig.update_traces(textposition='top center')
fig.update_layout(height=800)
fig.show()
```
